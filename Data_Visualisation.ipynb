```python
import pandas as pd

df = pd.read_csv("student_visualization_dataset (1).csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
import matplotlib.pyplot as plt

plt.plot(df["Student_ID"], df["Marks"])
plt.xlabel("Student_ID")
plt.ylabel("Marks")
plt.title("Line Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.hist(df["Marks"])
plt.xlabel("Marks")
plt.ylabel("Student_ID")
plt.title("RESULT ANALYSIS")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Study_Hours"], df["Marks"])
plt.xlabel("Study_Hours")
plt.ylabel("Marks")
plt.title("Positive Relation")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Screen_Time_Hours"], df["Marks"])
plt.xlabel("Screen_Time_Hours")
plt.ylabel("Sleep Hours")
plt.title("Negative Relation")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Marks"].value_counts().plot.pie(autopct="%1.1f%%")

plt.title("Attendance Distribution")
plt.ylabel("")
plt.show()
```
